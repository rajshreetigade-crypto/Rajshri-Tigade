**pd.get_dummies**

- get dummies is a way to represent the categorical columns to numerical=====>indepenent ==> 90degree
- first it will check what are the labels in a column
- for example case status has two columns
    - Denied and Certified
- so it will create two new columns based  on labels
   - col1:case_status_Denied
   - col2:case_status_Certified


case status | case_status_Denied | case_status_Certified
     -                -                     -
Denied      |          1         |           0
Certified   |          0         |           1
Certified   |          0         |           1


**It is also called as One Hot Encoder**
- It makes new columns are independent each other
- Some of ML algorithm expect that column should be **independent** to each other
